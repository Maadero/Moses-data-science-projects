```python
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold, cross_validate, RandomizedSearchCV
from sklearn.tree import DecisionTreeClassifier
from sklearn.impute import KNNImputer
from sklearn.feature_selection import RFE
from sklearn.metrics import recall_score, confusion_matrix, make_scorer, accuracy_score, f1_score
from sklearn.base import BaseEstimator, ClassifierMixin, clone
from sklearn.neighbors import NearestNeighbors
from scipy.stats import randint

# Ensure features and target are in pandas DataFrame format
def ensure_dataframe(X, y):
    """
    Ensure that the input features `X` and target `y` are in pandas DataFrame and Series format respectively.

    Parameters:
    -----------
    X : array-like or pandas DataFrame
        The input feature set.
    
    y : array-like, pandas Series, or pandas DataFrame
        The target labels.

    Returns:
    --------
    X : pandas DataFrame
        The input features converted to a pandas DataFrame.
    
    y : pandas Series
        The target labels converted to a pandas Series.
    """
    if not isinstance(X, pd.DataFrame):
        X = pd.DataFrame(X).reset_index(drop=True)
    if not isinstance(y, pd.Series):
        if isinstance(y, pd.DataFrame):
            y = y.iloc[:, 0].reset_index(drop=True)  # Extract the first column if y is a DataFrame
        else:
            y = pd.Series(y).reset_index(drop=True)
    return X, y

# Load datasets and ensure they are in the correct format
X_bcwd, y_bcwd = ensure_dataframe(X_bcwd, y_bcwd)
X_bcwo, y_bcwo = ensure_dataframe(X_bcwo, y_bcwo)

# Impute missing values
imputer = KNNImputer(n_neighbors=5)
X_bcwo = pd.DataFrame(imputer.fit_transform(X_bcwo), columns=X_bcwo.columns)

# Custom scorers for specificity and G-mean
def specificity_score(y_true, y_pred):
    """
    Calculates the specificity (true negative rate) of the predictions.

    Parameters:
    -----------
    y_true : array-like
        True binary labels.
    
    y_pred : array-like
        Predicted binary labels.

    Returns:
    --------
    specificity : float
        The specificity score.
    """
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
    return tn / (tn + fp)

def geometric_mean_score(y_true, y_pred):
    """
    Calculates the geometric mean of sensitivity (recall) and specificity.

    Parameters:
    -----------
    y_true : array-like
        True binary labels.
    
    y_pred : array-like
        Predicted binary labels.

    Returns:
    --------
    g_mean : float
        The geometric mean score.
    """
    sensitivity = recall_score(y_true, y_pred)
    specificity = specificity_score(y_true, y_pred)
    return np.sqrt(sensitivity * specificity)

specificity = make_scorer(specificity_score)
g_mean = make_scorer(geometric_mean_score, greater_is_better=True)
f1 = make_scorer(f1_score)
accuracy = make_scorer(accuracy_score)

# RAMOBoost Class Definition
class RAMOBoost(BaseEstimator, ClassifierMixin):
    """
    RAMOBoost is an ensemble learning method designed to handle class imbalance by generating synthetic minority samples 
    during training. The base estimator is trained iteratively on resampled data with a focus on improving minority class 
    predictions.

    Parameters:
    -----------
    base_estimator : estimator object, default=None
        The base estimator from which the boosted ensemble is built. If None, a DecisionTreeClassifier with max_depth=3 is used.
    
    n_estimators : int, default=30
        The number of boosting rounds.
    
    k1 : int, default=12
        Number of neighbors to consider in the rank calculation.
    
    k2 : int, default=9
        Number of neighbors used for generating synthetic samples.
    
    alpha : float, default=0.5
        The scaling factor for rank calculation.
    
    N : int, default=5
        The number of synthetic samples to generate per minority instance.
    
    random_state : int or None, default=None
        The random seed for reproducibility.
    
    Attributes:
    -----------
    estimators_ : list of estimator objects
        The collection of fitted base estimators.
    
    betas_ : list of floats
        The weights for each base estimator.
    
    sample_weights : array-like
        The sample weights at the end of the boosting process.
    """

    def __init__(self, base_estimator=None, n_estimators=30, k1=12, k2=9, alpha=0.5, N=5, random_state=None):
        self.base_estimator = base_estimator if base_estimator is not None else DecisionTreeClassifier(max_depth=3)
        self.n_estimators = n_estimators
        self.k1 = k1
        self.k2 = k2
        self.alpha = alpha
        self.N = N
        self.random_state = random_state
        self.estimators_ = []
        self.betas_ = []
        self.sample_weights = None

    def fit(self, X, y):
        """
        Trains the RAMOBoost model on the given dataset `X` and target `y`.

        Parameters:
        -----------
        X : pandas DataFrame or array-like
            The input feature set.
        
        y : pandas Series or array-like
            The target labels.

        Returns:
        --------
        self : object
            Returns the instance itself.
        """
        if not isinstance(X, pd.DataFrame):
            X = pd.DataFrame(X)
        if not isinstance(y, pd.Series):
            y = pd.Series(y)
        
        X = X.reset_index(drop=True)
        y = y.reset_index(drop=True)
        
        m, n = X.shape
        B = np.array([(i, y[i]) for i in range(m)])
        self.sample_weights = np.ones(len(B)) / len(B)
        
        for t in range(self.n_estimators):
            Se = self._sample_with_distribution(X, y, self.sample_weights)
            e1, e2 = self._split_majority_minority(Se)
            
            # Ranking minority instances
            ranks = self._calculate_ranks(e2, e1)
            
            # Sampling minority instances based on ranks
            gt = self._sample_minority(e2, ranks)
            
            # Adjust k2 if necessary
            k2_effective = min(self.k2, len(gt))
            
            # Generating synthetic samples (less aggressive)
            synthetic_samples = self._generate_synthetic_samples(gt, k2_effective)
            
            # Check if synthetic samples are generated
            if synthetic_samples.size > 0:
                Se_with_synthetic = np.vstack([Se, synthetic_samples])
            else:
                Se_with_synthetic = Se
            
            # Train base classifier
            h_t = clone(self.base_estimator).fit(Se_with_synthetic[:, :-1], Se_with_synthetic[:, -1])
            self.estimators_.append(h_t)
            
            # Update weights with penalty for over-predicting minority class
            epsilon_t, Z_t = self._calculate_pseudo_loss(h_t, X, y, self.sample_weights)
            beta_t = epsilon_t / (1 - epsilon_t)
            self.betas_.append(beta_t)
            
            self.sample_weights = self._update_weights(self.sample_weights, h_t, X, y, beta_t, Z_t)
        return self

    def _sample_with_distribution(self, X, y, D):
        """
        Samples instances from the dataset `X` and `y` based on the distribution `D`.

        Parameters:
        -----------
        X : array-like
            The input feature set.
        
        y : array-like
            The target labels.
        
        D : array-like
            The sampling distribution.

        Returns:
        --------
        Se : array-like
            The sampled dataset with the last column as the target labels.
        """
        X = X.values if isinstance(X, pd.DataFrame) else X
        y = y.values if isinstance(y, pd.Series) else y
        D = D / np.sum(D)
        indices = np.random.choice(np.arange(len(D)), size=len(D), p=D)
        return np.hstack([X[indices], y[indices].reshape(-1, 1)])

    def _split_majority_minority(self, Se):
        """
        Splits the sampled instances into majority and minority classes.

        Parameters:
        -----------
        Se : array-like
            The sampled dataset with the last column as the target labels.

        Returns:
        --------
        e1 : array-like
            The majority class instances.
        
        e2 : array-like
            The minority class instances.
        """
        e1 = Se[Se[:, -1] == 0]
        e2 = Se[Se[:, -1] == 1]
        return e1, e2

    def _calculate_ranks(self, e2, e1):
        """
        Calculates ranks for minority instances based on their proximity to majority instances.

        Parameters:
        -----------
        e2 : array-like
            The minority class instances.
        
        e1 : array-like
            The majority class instances.

        Returns:
        --------
        ranks : array-like
            The calculated ranks for the minority instances.
        """
        ranks = []
        for instance in e2:
            neighbors = NearestNeighbors(n_neighbors=min(self.k1, len(e1))).fit(e1[:, :-1])
            distances, indices = neighbors.kneighbors(instance[:-1].reshape(1, -1))
            delta_i = len(indices[indices == 0])
            r_i = 1 / (1 + np.exp(-self.alpha * delta_i))
            ranks.append(r_i)
        ranks = np.array(ranks)
        ranks /= np.sum(ranks)
        return ranks

    def _sample_minority(self, e2, ranks):
        """
        Samples minority instances according to their calculated ranks.

        Parameters:
        -----------
        e2 : array-like
            The minority class instances.
        
        ranks : array-like
            The calculated ranks for the minority instances.

        Returns:
        --------
        gt : array-like
            The sampled minority instances.
        """
        indices = np.random.choice(np.arange(len(ranks)), size=len(ranks), p=ranks)
        return e2[indices]

    def _generate_synthetic_samples(self, gt, k2):
        """
        Generates synthetic minority samples using the nearest neighbors of the sampled minority instances.

        Parameters:
        -----------
        gt : array-like
            The sampled minority instances.
        
        k2 : int
            The number of neighbors to consider for generating synthetic samples.

        Returns:
        --------
        synthetic_samples : array-like
            The generated synthetic samples.
        """
        synthetic_samples = []
        for instance in gt:
            neighbors = NearestNeighbors(n_neighbors=k2).fit(gt[:, :-1])
            distances, indices = neighbors.kneighbors(instance[:-1].reshape(1, -1))
            for _ in range(self.N // len(gt)):
                neighbor = gt[np.random.choice(indices.flatten())]
                synthetic = instance[:-1] + np.random.random() * (neighbor[:-1] - instance[:-1])
                synthetic_samples.append(np.hstack([synthetic, [1]]))
        return np.array(synthetic_samples)

    def _calculate_pseudo_loss(self, h_t, X, y, D):
        """
        Calculates the pseudo-loss for the current base estimator.

        Parameters:
        -----------
        h_t : estimator object
            The current base estimator.
        
        X : array-like
            The input feature set.
        
        y : array-like
            The target labels.
        
        D : array-like
            The current sample weights.

        Returns:
        --------
        epsilon_t : float
            The pseudo-loss of the current base estimator.
        
        Z_t : float
            The normalization factor for sample weights.
        """
        h_pred = h_t.predict(X)
        epsilon_t = 0.5 * np.sum(D * (1 - (y == h_pred).astype(int)))
        Z_t = np.sum(D)
        return epsilon_t, Z_t

    def _update_weights(self, D, h_t, X, y, beta_t, Z_t):
        """
        Updates the sample weights based on the performance of the current base estimator.

        Parameters:
        -----------
        D : array-like
            The current sample weights.
        
        h_t : estimator object
            The current base estimator.
        
        X : array-like
            The input feature set.
        
        y : array-like
            The target labels.
        
        beta_t : float
            The weight of the current base estimator.
        
        Z_t : float
            The normalization factor for sample weights.

        Returns:
        --------
        D : array-like
            The updated sample weights.
        """
        h_pred = h_t.predict(X)
        penalty_factor = np.where(h_pred == 1, 0.5, 1.0)
        D = D * np.exp(beta_t * penalty_factor * (1 - (y == h_pred).astype(int)))
        return D / Z_t

    def predict(self, X):
        """
        Predicts class labels for the input dataset `X` by combining predictions from all base estimators.

        Parameters:
        -----------
        X : pandas DataFrame or array-like
            The input feature set.

        Returns:
        --------
        predictions : array-like
            The predicted class labels.
        """
        final_prediction = np.zeros(X.shape[0])
        for beta_t, h_t in zip(self.betas_, self.estimators_):
            final_prediction += np.log(1 / beta_t) * h_t.predict(X)
        return np.sign(final_prediction)

    def predict_proba(self, X):
        """
        Predicts class probabilities for the input dataset `X` by averaging the probabilities from all base estimators.

        Parameters:
        -----------
        X : pandas DataFrame or array-like
            The input feature set.

        Returns:
        --------
        probabilities : array-like
            The predicted class probabilities.
        """
        final_proba = np.zeros(X.shape[0])
        for beta_t, h_t in zip(self.betas_, self.estimators_):
            final_proba += np.log(1 / beta_t) * h_t.predict_proba(X)[:, 1]
        return final_proba / np.sum([np.log(1 / beta_t) for beta_t in self.betas_])

    def get_params(self, deep=True):
        """
        Returns the parameters of the model.

        Parameters:
        -----------
        deep : bool, default=True
            If True, will return the parameters for this estimator and contained subobjects that are estimators.

        Returns:
        --------
        params : dict
            Parameter names mapped to their values.
        """
        return super().get_params(deep)

    def set_params(self, **params):
        """
        Sets the parameters of the model.

        Parameters:
        -----------
        **params : dict
            The parameter names mapped to their new values.

        Returns:
        --------
        self : object
            Returns the instance itself.
        """
        return super().set_params(**params)



n_features_values = {
    'BCWD': (10, 30),
    'BCWO': (3, 9),
    'BCT': (10, 38)
}

datasets = {
    'BCWD': (X_bcwd, y_bcwd),
    'BCWO': (X_bcwo, y_bcwo),
    'BCT': (X_bct, y_bct)
}

results = []

for dataset_name, (X, y) in datasets.items():
    print(f"Running RAMOBoost on {dataset_name} dataset...")

    dt = DecisionTreeClassifier(random_state=101)
    rfe = RFE(estimator=dt, n_features_to_select=n_features_values[dataset_name][0])  # Start with the lower bound
    X_reduced = rfe.fit_transform(X, y)

    ramoboost = RAMOBoost(
        base_estimator=DecisionTreeClassifier(random_state=101), 
        n_estimators=20, 
        k1=5, 
        k2=5, 
        alpha=0.001, 
        N=5, 
        random_state=101
    )
    
    param_distributions = {
        'base_estimator__max_depth': randint(3, 7),
        'n_estimators': randint(10, 30),
        'k1': randint(1, 20),
        'k2': randint(1, 20),
        'alpha': np.logspace(-3, 0, num=100),
        'N': randint(1, 10)
    }

    stratified_kfold = StratifiedKFold(n_splits=10, shuffle=True, random_state=101)

    random_search = RandomizedSearchCV(
        estimator=ramoboost,
        param_distributions=param_distributions,
        n_iter=50,
        cv=stratified_kfold,
        scoring={'accuracy': 'accuracy', 'sensitivity': 'recall', 'specificity': specificity, 'g_mean': g_mean, 'f1': f1},
        refit='g_mean',
        n_jobs=-1,
        random_state=101
    )

    random_search.fit(X_reduced, y)

    cv_results = cross_validate(
        random_search.best_estimator_, 
        X_reduced, 
        y, 
        cv=stratified_kfold, 
        scoring={
            'accuracy': 'accuracy',
            'sensitivity': 'recall',
            'specificity': specificity,
            'g_mean': g_mean,
            'f1': f1
        }, 
        n_jobs=-1, 
        return_train_score=False
    )

    result = {
        'Dataset': dataset_name,
        'Method': 'RAMOBoost With Feature Selection',
        'Best Params': random_search.best_params_,
        'Selected Features': rfe.n_features_,
        'Accuracy': np.mean(cv_results['test_accuracy']),
        'Sensitivity': np.mean(cv_results['test_sensitivity']),
        'Specificity': np.mean(cv_results['test_specificity']),
        'G-Mean': np.mean(cv_results['test_g_mean']),
        'F1': np.mean(cv_results['test_f1'])
    }
    results.append(result)

    # run without feature selection
    random_search.fit(X, y)

    cv_results_no_fs = cross_validate(
        random_search.best_estimator_, 
        X, 
        y, 
        cv=stratified_kfold, 
        scoring={
            'accuracy': 'accuracy',
            'sensitivity': 'recall',
            'specificity': specificity,
            'g_mean': g_mean,
            'f1': f1
        }, 
        n_jobs=-1, 
        return_train_score=False
    )

    # Store the results without feature selection
    result_no_fs = {
        'Dataset': dataset_name,
        'Method': 'RAMOBoost Without Feature Selection',
        'Best Params': random_search.best_params_,
        'Selected Features': X.shape[1],  # All features are used
        'Accuracy': np.mean(cv_results_no_fs['test_accuracy']),
        'Sensitivity': np.mean(cv_results_no_fs['test_sensitivity']),
        'Specificity': np.mean(cv_results_no_fs['test_specificity']),
        'G-Mean': np.mean(cv_results_no_fs['test_g_mean']),
        'F1': np.mean(cv_results_no_fs['test_f1'])
    }
    results.append(result_no_fs)

results_df = pd.DataFrame(results).round(4)
results_df.to_csv('results_ramoboost_all_datasets.csv', index=False)
print("Final results across all datasets:")
print(results_df)
```
